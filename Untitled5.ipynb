# 7. Normalización de las Señales


Aplicamos una normalización estándar (media cero y desviación estándar uno) a las señales EEG.


```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(x_train_df)
x_train_df.describe().T
```

# 8. Extracción de características con PyTS o TsFresh.

```python

y_train_series = pd.Series(y_train, name="target")

idx_class_0 = y_train_series[y_train_series == 0].sample(50, random_state=42).index
idx_class_1 = y_train_series[y_train_series == 1].sample(50, random_state=42).index
idx_sample = np.concatenate([idx_class_0, idx_class_1])

X_sample = X_train[idx_sample]  
y_sample = y_train_series[idx_sample].reset_index(drop=True)

def prepare_tsfresh_input(X, ids):
    records = []
    for i, sample_id in enumerate(ids):
        for channel in range(X.shape[1]):
            values = X[i, channel, :]
            for t, val in enumerate(values):
                records.append((sample_id * 64 + channel, t, val))
    df = pd.DataFrame(records, columns=["id", "time", "value"])
    return df.dropna()  

df_long = prepare_tsfresh_input(X_sample, np.arange(len(X_sample)))

X_features = extract_features(df_long,
                              column_id="id",
                              column_sort="time",
                              default_fc_parameters=EfficientFCParameters(),  # Funciones rápidas
                              n_jobs=4)




```

```python
X_features = impute(X_features)
```

```python
y_repeated = y_sample.repeat(64).reset_index(drop=True)
X_features = X_features.reset_index(drop=True)
y_repeated_aligned = y_repeated.loc[X_features.index].reset_index(drop=True)

X_selected = select_features(X_features, y_repeated_aligned)

print("Forma final de las características seleccionadas:", X_selected.shape)
```

```python
print("Distribución de clases:", y_repeated.value_counts())
```

# 8. PCA (Análisis de Componentes Principales)


Aplicamos PCA para reducir la dimensionalidad de las señales y ver cuántos componentes necesitamos para explicar el 95% de la varianza

```python
pca = PCA()
pca.fit(X_train_scaled)

# Calcular varianza explicada acumulada
explained_variance = np.cumsum(pca.explained_variance_ratio_)

# Graficar varianza explicada
plt.plot(explained_variance)
plt.xlabel('Número de componentes')
plt.ylabel('Varianza explicada acumulada')
plt.title('Varianza explicada acumulada por PCA')
plt.grid(True)
plt.show()

# ¿Cuántos componentes necesitamos para el 95% de varianza?
n_components = np.argmax(explained_variance >= 0.95) + 1  # +1 para contar desde 1
print(f"Componentes necesarios para el 95% de la varianza: {n_components}")
```

# 9. Reducir la Dimensionalidad con PCA


```python
pca = PCA(n_components=n_components)
X_train_pca = pca.fit_transform(X_train_scaled)

print(f"Forma de los datos después de PCA: {X_train_pca.shape}")
```

# 10. Entrenamiento del Modelo SVM


Entrenamos un modelo SVM con kernel lineal y evaluamos su rendimiento.


```python
X_train_pca, X_val_pca, y_train_pca, y_val_pca = train_test_split(X_train_pca, y_train, test_size=0.2, random_state=42)

model = SVC(kernel='linear')
model.fit(X_train_pca, y_train_pca)

y_pred_pca = model.predict(X_val_pca)

print("Reporte de clasificación:\n", classification_report(y_val_pca, y_pred_pca))
print("Matriz de confusión:\n", confusion_matrix(y_val_pca, y_pred_pca))
print(f"Precisión: {accuracy_score(y_val_pca, y_pred_pca)}")
```
